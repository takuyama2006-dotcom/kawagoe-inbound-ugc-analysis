```python
!pip install -q japanize-matplotlib

import os
import matplotlib.pyplot as plt
import pandas as pd
import japanize_matplotlib
from ipywidgets import interact
summary_data = {
    '言語': ['日本語', '中国語', '英語', '韓国語'],
    '動画本数': [366, 384, 414, 367],
    '平均再生回数': [29242.1, 25240.7, 4398.7, 7106.2],
    '再生数の中央値': [3361.5, 3598.5, 322.0, 626.0],
    '平均高評価率(%)': [1.27, 1.55, 3.68, 2.89],
    '平均コメント率(%)': [0.13, 0.17, 0.80, 0.85],
}
df_summary = pd.DataFrame(summary_data)

print("==========================================")
print(" 4言語別 全体パフォーマンス指標比較表")
print("==========================================")
display(df_summary)


print("\n" + "=" * 50 + "\n")
files = {
    '日本語': 'kawagoe_ja.csv',
    '中国語': 'kawagoe_ch.csv',
    '英語': 'kawagoe_en.csv',
    '韓国語': 'kawagoe_ko.csv',
}

def analyze_kawagoe(言語):
    filepath = files[言語]

    if not os.path.exists(filepath):
        print(
            f"❌ ファイル '{filepath}' が見つかりません。左パネルの「ファイル」にアップロードされているか確認してください。"
        )
        return

    df = pd.read_csv(filepath)

    print(f"===【{言語}】 キーワード別分析結果 ===")

    df_sorted = df.sort_values(by='平均再生回数', ascending=False)

    display_cols = [
        col
        for col in [
            'キーワード',
            '検索語',
            '動画本数',
            '平均再生回数',
            '再生数の中央値',
            '平均高評価数'
            '平均高評価率',
            '平均コメント率',
        ]
        if col in df_sorted.columns
    ]
    display(df_sorted[display_cols].reset_index(drop=True))

    plt.figure(figsize=(10, 5))
    plt.barh(
        df_sorted['キーワード'],
        df_sorted['平均再生回数'],
        color='steelblue',
    )
    plt.xlabel('平均再生回数')
    plt.ylabel('キーワード')
    plt.title(f'【{言語}】 キーワード別 平均再生回数比較')
    plt.gca().invert_yaxis()
    plt.grid(axis='x', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()
    df_sorted_med = df.sort_values(by='再生数の中央値', ascending=False)
    plt.figure(figsize=(10, 5))
    plt.barh(
        df_sorted_med['キーワード'],
        df_sorted_med['再生数の中央値'],
        color='steelblue',
    )
    plt.xlabel('再生数の中央値')
    plt.ylabel('キーワード')
    plt.title(f'【{言語}】 キーワード別 再生数の中央値比較')
    plt.gca().invert_yaxis()
    plt.grid(axis='x', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()
    df_sorted_med = df.sort_values(by='平均高評価数', ascending=False)
    plt.figure(figsize=(10, 5))
    plt.barh(
        df_sorted_med['キーワード'],
        df_sorted_med['平均高評価数'],
        color='steelblue',
    )
    plt.xlabel('平均高評価数')
    plt.ylabel('キーワード')
    plt.title(f'【{言語}】 キーワード別 平均高評価数比較')
    plt.gca().invert_yaxis()
    plt.grid(axis='x', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()
    df_sorted_high = df.sort_values(by='平均高評価率', ascending=False)
    plt.figure(figsize=(10, 5))
    plt.barh(
        df_sorted_high['キーワード'],
        df_sorted_high['平均高評価率'],
        color='steelblue',
    )
    plt.xlabel('平均高評価率')
    plt.ylabel('キーワード')
    plt.title(f'【{言語}】 キーワード別 平均高評価率比較')
    plt.gca().invert_yaxis()
    plt.grid(axis='x', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()
    df_sorted_med = df.sort_values(by='平均コメント率', ascending=False)
    plt.figure(figsize=(10, 5))
    plt.barh(
        df_sorted_med['キーワード'],
        df_sorted_med['平均コメント率'],
        color='steelblue',
    )
    plt.xlabel('平均コメント率')
    plt.ylabel('キーワード')
    plt.title(f'【{言語}】 キーワード別 平均コメント率比較')
    plt.gca().invert_yaxis()
    plt.grid(axis='x', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()

interact(analyze_kawagoe, 言語=['日本語', '中国語', '英語', '韓国語'])
```
